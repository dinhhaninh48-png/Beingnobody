```python
import os
import sys
import time
import subprocess
import threading
import psutil
from pathlib import Path

MODEL_DIR = "/content/models/VoxCPM2"
OUTPUT = "/content/male_latenight_baseline.wav"
LOG = "/content/voxcpm_diagnostic.log"

CONTROL = (
    "Mature Vietnamese male narrator, medium-low pitch, "
    "warm deep and smooth voice, calm and gentle with a "
    "reflective tone, slow peaceful pace, clear soft articulation, "
    "intimate late-night storytelling delivery, "
    "soothing and emotionally expressive without sounding theatrical."
)

TEXT = (
    "Chào bạn, cảm ơn bạn đã ở đây giờ phút này. "
    "Hãy hít một hơi thật sâu, thả lỏng bờ vai "
    "và để tâm trí được nghỉ ngơi sau một ngày dài nhé..."
)

env = os.environ.copy()
env["PYTHONPATH"] = "/content/Beingnobody/src"
env["HF_HUB_OFFLINE"] = "1"
env["PYTHONUNBUFFERED"] = "1"

cmd = [
    sys.executable, "-m", "voxcpm.cli", "design",
    "--model-path", MODEL_DIR,
    "--device", "cuda",
    "--local-files-only",
    "--no-denoiser",
    "--no-optimize",
    "--control", CONTROL,
    "--text", TEXT,
    "--seed", "42",
    "--cfg-value", "2.0",
    "--inference-timesteps", "10",
    "--output", OUTPUT,
]

stop = threading.Event()
samples = []

def monitor(proc):
    child = psutil.Process(proc.pid)

    while not stop.is_set():
        try:
            ram = psutil.virtual_memory()

            rss = child.memory_info().rss / 2**30

            gpu = subprocess.run(
                [
                    "nvidia-smi",
                    "--query-gpu=memory.used,memory.free",
                    "--format=csv,noheader,nounits",
                ],
                capture_output=True,
                text=True,
                timeout=5,
            )

            sample = {
                "time": round(time.time(), 1),
                "system_available_gib": round(ram.available / 2**30, 2),
                "process_rss_gib": round(rss, 2),
                "gpu_memory_mib": gpu.stdout.strip(),
            }

            samples.append(sample)
            print(sample, flush=True)

        except Exception as e:
            print("Monitor:", type(e).__name__, str(e))

        stop.wait(2)

start = time.perf_counter()

with open(LOG, "w", encoding="utf-8") as log:
    proc = subprocess.Popen(
        cmd,
        env=env,
        stdout=log,
        stderr=subprocess.STDOUT,
    )

    worker = threading.Thread(
        target=monitor,
        args=(proc,),
        daemon=True,
    )
    worker.start()

    try:
        exit_code = proc.wait(timeout=300)
    except subprocess.TimeoutExpired:
        proc.kill()
        exit_code = proc.wait()
        print("Diagnostic timeout after 300 seconds")
    finally:
        stop.set()
        worker.join(timeout=10)

print("\n=== RESULT ===")
print("Exit code:", exit_code)
print("Elapsed:", round(time.perf_counter() - start, 1), "seconds")
print("Output exists:", Path(OUTPUT).is_file())

print("\n=== LAST LOG LINES ===")
if Path(LOG).exists():
    print(Path(LOG).read_text(errors="replace")[-6000:])

print("\n=== MEMORY PEAKS ===")
if samples:
    print(
        "Peak process RSS:",
        max(s["process_rss_gib"] for s in samples),
        "GiB"
    )
    print(
        "Minimum available RAM:",
        min(s["system_available_gib"] for s in samples),
        "GiB"
    )
```

```python
import shutil
import hashlib
from pathlib import Path

def save_wav_to_drive(wav_path, destination_dir=DRIVE_OUTPUT_DIR):
    source = Path(wav_path)
    destination_dir = Path(destination_dir)

    if not source.is_file() or source.stat().st_size == 0:
        raise FileNotFoundError(f"WAV không tồn tại hoặc rỗng: {source}")

    destination_dir.mkdir(parents=True, exist_ok=True)
    target = destination_dir / source.name

    # Không ghi đè file đã tồn tại
    if target.exists():
        raise FileExistsError(f"File đã tồn tại trên Drive: {target}")

    shutil.copy2(source, target)

    def sha256(path):
        h = hashlib.sha256()
        with open(path, "rb") as f:
            for chunk in iter(lambda: f.read(1024 * 1024), b""):
                h.update(chunk)
        return h.hexdigest()

    assert target.is_file()
    assert sha256(source) == sha256(target), "Checksum không khớp"

    print("DRIVE SAVE VERIFIED")
    print("Saved:", target)
    return target
```

```python

import os
import sys
import json
import time
import hashlib
import subprocess
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import soundfile as sf

# ============================================================
# 1. CONFIGURATION
# ============================================================

PRESET_ID = "vi_male_latenight_01"
MODEL_REVISION = "32279effe8c19989596f05d353d1447f51d9e915"

MODEL_DIR = Path("/content/models/VoxCPM2")
SOURCE = Path("/content/Beingnobody/src/voxcpm/model/voxcpm2.py")
DRIVE_ROOT = Path(
    "/content/drive/MyDrive/AI_LAB/VoxCPM2/outputs/cua_so_thien_tap_01"
)

MAX_CHARS = 420
PAUSE_SECONDS = 0.45
SEED = 42
CFG = 2.0
TIMESTEPS = 10

CONTROL = (
    "Mature Vietnamese male narrator, medium-low pitch, "
    "warm deep and smooth voice, calm and gentle with a "
    "reflective tone, slow peaceful pace, clear soft "
    "articulation, intimate late-night storytelling "
    "delivery, soothing and emotionally expressive "
    "without sounding theatrical."
)

# ============================================================
# 2. COMPLETE VOICEOVER SCRIPT
# Each non-empty line is treated as a sentence/reading unit.
# ============================================================

SCRIPT = """
Có những chuyện xảy ra trong đời mà mình không được quyền chọn.
Một người làm mình tổn thương.
Một lần thất bại.
Một cuộc chia xa.
Một điều quen thuộc bỗng thay đổi.
Nhưng sau tất cả những chuyện mình không được chọn ấy…
vẫn còn một lựa chọn rất quan trọng.
Mình sẽ trở thành người như thế nào sau khi đi qua chúng?

“Cuộc Sống Nhìn Từ Ô Cửa Thiền – Tập 1” của Vô Thường khiến tôi nghĩ nhiều nhất về câu hỏi ấy.
Cuốn sách không chỉ nói về buông bỏ.
Nó đi qua tổn thương, sợ hãi, oán giận, tình thương, đổi thay…
và cách một người sống tiếp khi cuộc đời không đi theo ý mình.

Có những lúc mình nghĩ bình yên chỉ đến khi mọi thứ bên ngoài ổn trước.
Công việc ổn.
Mối quan hệ ổn.
Người khác hiểu mình.
Những chuyện cũ không còn làm đau.
Nhưng cuộc sống hiếm khi đứng yên để đợi mình bình an.
Một chuyện vừa qua…
thì chuyện khác lại đến.
Một người bước vào…
rồi cũng có lúc rời đi.
Một điều tưởng rất chắc chắn…
rồi một ngày đổi khác.
Có lẽ điều khó không phải là tìm một cuộc đời không biến động.
Mà là học cách sống trong một cuộc đời luôn biến động.

Ở những phần đầu của cuốn sách, một điều xuất hiện khá rõ là cách con người mang theo nỗi đau.
Có những chuyện chỉ xảy ra một lần.
Nhưng trong tâm trí, mình có thể sống lại nó rất nhiều lần.
Một lời nói.
Một lần bị bỏ rơi.
Một điều không công bằng.
Mình nhớ lại.
Rồi giận.
Rồi tự hỏi tại sao.
Có lúc mình tưởng rằng mình đang giữ ký ức.
Nhưng thật ra…
ký ức đang giữ mình.
Đây là suy ngẫm của người kể từ tinh thần phần sách đã đọc, không phải câu nguyên văn của tác giả.

Nhưng điều khiến cuốn sách đi xa hơn ý niệm “buông xuống để bình yên”…
là ở chỗ nó không chỉ hỏi mình đang giữ điều gì.
Nó còn khiến mình nhìn vào một câu hỏi khó hơn:
Những điều đã làm mình đau…
đang biến mình thành người như thế nào?

Một người từng bị tổn thương có thể trở nên cứng hơn.
Một người từng bị bỏ rơi có thể bắt đầu sợ gần gũi.
Một người từng bị phản bội có thể trở nên nghi ngờ tất cả.
Và rồi một vết thương không còn nằm ở một người nữa.
Nó được truyền tiếp.
Từ một lời nói.
Sang một thái độ.
Từ một người.
Sang một người khác.
Đó là lúc câu chuyện không còn chỉ là chữa lành cho mình.
Mà còn là:
mình có tiếp tục trao nỗi đau này cho người khác không?

Có những người sau khi biết đau…
sẽ làm người khác đau để tự bảo vệ mình.
Nhưng cũng có những người…
chính vì biết đau là gì,
nên cẩn trọng hơn với nỗi đau của người khác.
Có lẽ đó là một dạng trưởng thành rất khó.
Không phải chưa từng đi qua bóng tối.
Mà đã đi qua…
nhưng không mang bóng tối ấy phát cho những người mình gặp sau này.

Cuốn sách cũng nhiều lần quay về sự thay đổi.
Những con đường bằng phẳng rồi có thể gập ghềnh.
Những điều từng rất gần rồi có thể trở nên xa.
Những người ta nghĩ sẽ ở lại…
không phải ai cũng ở lại.
Thời gian không dừng.
Tuổi tác không dừng.
Một ngày cũng không dừng.
Và điều này nghe có vẻ buồn.
Nhưng vô thường không chỉ có nghĩa là mất đi.
Bởi nếu mọi thứ đều không thay đổi…
thì nỗi đau cũng sẽ không bao giờ thay đổi.
Một con người cũng không thể khác đi.
Một ngày mới cũng không thể bắt đầu.

Chính vì mọi thứ chuyển động…
mình mới có cơ hội bước khỏi một phiên bản cũ của mình.
Một điều từng rất đau…
có thể một ngày không còn đau như trước.
Một thất bại từng khiến mình nghĩ tất cả đã kết thúc…
có thể vài năm sau chỉ còn là một đoạn đường.
Một người từng nghĩ mình không thể sống thiếu ai đó…
rồi vẫn tiếp tục sống.
Không phải vì những gì đã xảy ra không quan trọng.
Mà vì con người cũng thay đổi cùng thời gian.

Có lẽ vì vậy, vô thường không chỉ là điều mình phải chấp nhận.
Nó cũng cho mình một cơ hội.
Cơ hội sửa lại.
Cơ hội học.
Cơ hội nhìn một việc khác đi.
Cơ hội không mãi là người mình từng là.

Nhưng thay đổi không tự động làm một người tốt hơn.
Thời gian trôi cũng không có nghĩa mọi vết thương đều được chữa lành.
Có khi nhiều năm đã qua…
mà trong lòng mình vẫn ở nguyên chỗ cũ.
Vì vậy, giữa những gì đang đổi thay quanh mình…
vẫn có một phần đòi hỏi lựa chọn.
Mình chọn giữ điều gì.
Mình chọn đặt xuống điều gì.
Mình chọn nói lời nào.
Mình chọn tha thứ hay tiếp tục nuôi oán giận.
Mình chọn sống chân thành…
hay để hoàn cảnh quyết định mình sẽ trở thành ai.

Và ở đây, “buông xuống” vẫn quan trọng.
Nhưng nó không phải toàn bộ câu chuyện.
Mình đặt xuống…
để còn chỗ cho một điều khác.
Đặt xuống một nỗi giận…
để còn chỗ cho sự bình tĩnh.
Đặt xuống một điều không thể thay đổi…
để còn chỗ cho hiện tại.
Đặt xuống một mong cầu đã làm mình kiệt sức…
để nhìn rõ điều gì vẫn còn ở trước mặt.
Buông không phải để trở thành một người không cảm xúc.
Mà để một cảm xúc không quyết định hết con người mình.

Ở phần sau của cuốn sách, khi những suy ngẫm mở rộng sang thiên nhiên, ngày tháng và những chuyển động rất bình thường của đời sống…
cảm giác “nhìn từ ô cửa thiền” trở nên rõ hơn.
Không nhất thiết là rời khỏi cuộc sống để nhìn nó từ xa.
Mà là dừng lại đủ lâu…
để nhìn một chuyện quen thuộc bằng một khoảng cách khác.
Một cơn mưa.
Một ngày.
Một con đường.
Một người bên cạnh.
Một chuyện vừa xảy ra.
Những thứ ấy có thể vẫn vậy.
Nhưng người nhìn thì đã khác.
Và đôi khi, chỉ cần góc nhìn thay đổi…
một điều từng chỉ là mất mát có thể trở thành bài học.
Một điều từng chỉ là thất bại có thể trở thành hướng rẽ.
Một điều từng chỉ là tổn thương…
có thể trở thành lý do để mình sống tử tế hơn.
Đây cũng là suy ngẫm của người kể, không phải câu trích trong sách.

Có lẽ vì vậy, nếu phải gói lại tinh thần tôi nhận được từ Tập 1 trong một ý…
thì đó không chỉ là:
hãy buông xuống để bình yên.
Mà là:
giữa tất cả những gì thay đổi…
hãy để ý xem mình đang trở thành ai.

Bởi mình không kiểm soát được hết những gì sẽ đến.
Không kiểm soát được người khác.
Không giữ được mọi điều mình thương.
Không ngăn được thời gian.
Nhưng mình vẫn còn một phần trách nhiệm rất lớn với chính mình.
Đó là cách mình đi qua những điều ấy.
Mình có thể mang theo oán giận.
Hoặc mang theo hiểu biết.
Có thể trở nên lạnh lùng.
Hoặc sâu sắc hơn.
Có thể để một vết thương định nghĩa mình.
Hoặc để nó nhắc mình đừng tạo thêm một vết thương khác.
Có thể mất một điều.
Nhưng không nhất thiết phải mất luôn phần tốt đẹp trong mình.

Nếu đang ở một giai đoạn mà cuộc sống có quá nhiều thứ không như ý…
có lẽ “Cuộc Sống Nhìn Từ Ô Cửa Thiền – Tập 1” không phải cuốn để đọc thật nhanh.
Có thể chỉ cần vài trang.
Rồi dừng lại.
Không phải để tìm ngay một lời giải.
Mà để thử nhìn lại chính mình.
Có điều gì mình vẫn đang ôm quá chặt?
Có vết thương nào mình đang mang từ quá khứ vào hiện tại?
Có nỗi đau nào mình vô tình truyền sang một người khác?
Có điều gì đã thay đổi…
nhưng mình vẫn chưa chịu thay đổi cùng?
Và có một câu hỏi đáng giữ lâu hơn tất cả:
Sau những gì đã đi qua…
mình muốn trở thành người như thế nào?

Có những điều rồi mình sẽ phải đặt xuống.
Có những vết thương rồi sẽ thành một phần quá khứ.
Có những người sẽ đến.
Có những người sẽ đi.
Có những ngày mình từng nghĩ không thể bước qua…
rồi cuối cùng cũng nằm lại phía sau.
Cuộc đời vẫn thay đổi.
Có lẽ điều mình có thể gìn giữ không phải là mọi thứ quanh mình.
Mà là…
con người mình chọn trở thành giữa tất cả những đổi thay ấy.
""".strip()

# ============================================================
# 3. PRE-FLIGHT CHECKS
# ============================================================

import torch

assert torch.cuda.is_available(), "CUDA không hoạt động"
assert MODEL_DIR.joinpath("config.json").is_file(), "Model missing"
assert SOURCE.is_file(), "Repository missing"
assert Path("/content/drive/MyDrive").is_dir(), "Drive not mounted"
assert (
    "original_default_dtype = torch.get_default_dtype()"
    in SOURCE.read_text(encoding="utf-8")
), "BF16 patch missing"

# ============================================================
# 4. SPLIT SCRIPT WITHOUT REMOVING WORDS
# ============================================================

units = [line.strip() for line in SCRIPT.splitlines() if line.strip()]

chunks = []
current = []

for unit in units:
    candidate = " ".join(current + [unit])

    if current and len(candidate) > MAX_CHARS:
        chunks.append(" ".join(current))
        current = [unit]
    else:
        current.append(unit)

if current:
    chunks.append(" ".join(current))

assert chunks, "Empty script"
assert " ".join(chunks) == " ".join(units), "Text mismatch"

print("Total reading units:", len(units))
print("Total characters:", len(SCRIPT))
print("Total chunks:", len(chunks))
print("Longest chunk:", max(map(len, chunks)))

# ============================================================
# 5. VERSIONED DRIVE OUTPUT
# ============================================================

job_config = {
    "preset": PRESET_ID,
    "revision": MODEL_REVISION,
    "script": SCRIPT,
    "control": CONTROL,
    "max_chars": MAX_CHARS,
    "pause_seconds": PAUSE_SECONDS,
    "seed": SEED,
    "cfg": CFG,
    "timesteps": TIMESTEPS,
}

job_hash = hashlib.sha256(
    json.dumps(
        job_config, ensure_ascii=False, sort_keys=True
    ).encode("utf-8")
).hexdigest()[:16]

RUN_DIR = DRIVE_ROOT / "runs" / job_hash
PARTS_DIR = RUN_DIR / "parts"
LOCAL_DIR = Path("/content/voxcpm_voiceover") / job_hash

PARTS_DIR.mkdir(parents=True, exist_ok=True)
LOCAL_DIR.mkdir(parents=True, exist_ok=True)

def checksum(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

def valid_wav(path):
    path = Path(path)
    if not path.is_file() or path.stat().st_size == 0:
        return False
    try:
        info = sf.info(str(path))
        return info.frames > 0 and info.samplerate > 0
    except Exception:
        return False

# Preserve job identity for resuming later.
manifest_path = RUN_DIR / "job.json"

if manifest_path.exists():
    existing = json.loads(manifest_path.read_text(encoding="utf-8"))
    assert existing == job_config, "Existing job config mismatch"
else:
    with manifest_path.open("x", encoding="utf-8") as f:
        json.dump(job_config, f, ensure_ascii=False, indent=2)

env = os.environ.copy()
env["PYTHONPATH"] = "/content/Beingnobody/src"
env["HF_HUB_OFFLINE"] = "1"
env["PYTHONUNBUFFERED"] = "1"

# ============================================================
# 6. GENERATE EACH CHUNK AND SAVE IMMEDIATELY
# ============================================================

for index, chunk_text in enumerate(chunks, start=1):
    chunk_id = hashlib.sha256(
        chunk_text.encode("utf-8")
    ).hexdigest()[:12]

    filename = f"part_{index:03d}_{chunk_id}.wav"
    drive_part = PARTS_DIR / filename
    local_part = LOCAL_DIR / filename

    if valid_wav(drive_part):
        print(f"[{index}/{len(chunks)}] SKIP: already on Drive")
        continue

    # Stop rather than replace a questionable Drive file.
    if drive_part.exists():
        raise RuntimeError(f"Invalid existing chunk: {drive_part}")

    print(
        f"\n[{index}/{len(chunks)}] GENERATING "
        f"{len(chunk_text)} characters...",
        flush=True
    )

    local_part.unlink(missing_ok=True)

    command = [
        sys.executable, "-m", "voxcpm.cli", "design",
        "--model-path", str(MODEL_DIR),
        "--device", "cuda",
        "--local-files-only",
        "--no-denoiser",
        "--no-optimize",
        "--control", CONTROL,
        "--text", chunk_text,
        "--seed", str(SEED),
        "--cfg-value", str(CFG),
        "--inference-timesteps", str(TIMESTEPS),
        "--output", str(local_part),
    ]

    result = subprocess.run(
        command,
        env=env,
        check=False,
        stdout=subprocess.DEVNULL,
        stderr=subprocess.PIPE,
        text=True,
    )

    if result.returncode != 0:
        print(result.stderr[-5000:])
        raise RuntimeError(
            f"Chunk {index} failed: exit {result.returncode}. "
            "Stop here; rerun this cell to resume."
        )

    if not valid_wav(local_part):
        raise RuntimeError(f"Chunk {index}: output WAV invalid")

    # Copy with exclusive creation. Never overwrite.
    import shutil

    with local_part.open("rb") as src, drive_part.open("xb") as dst:
        shutil.copyfileobj(src, dst)

    if checksum(local_part) != checksum(drive_part):
        raise RuntimeError(
            f"Chunk {index}: Drive checksum mismatch"
        )

    print(
        f"[{index}/{len(chunks)}] DRIVE VERIFIED | "
        f"{sf.info(str(drive_part)).duration:.1f}s",
        flush=True,
    )

    local_part.unlink(missing_ok=True)

# ============================================================
# 7. JOIN WAV CHUNKS WITH SHORT PAUSES
# ============================================================

parts = []

for index, chunk_text in enumerate(chunks, start=1):
    chunk_id = hashlib.sha256(
        chunk_text.encode("utf-8")
    ).hexdigest()[:12]

    part = PARTS_DIR / f"part_{index:03d}_{chunk_id}.wav"
    assert valid_wav(part), f"Missing or invalid: {part}"
    parts.append(part)

first_info = sf.info(str(parts[0]))
sr = first_info.samplerate
channels = first_info.channels

for part in parts:
    info = sf.info(str(part))
    assert info.samplerate == sr, f"Sample rate mismatch: {part}"
    assert info.channels == channels, f"Channels mismatch: {part}"

final_local = LOCAL_DIR / "cua_so_thien_tap_01_full.wav"
final_local.unlink(missing_ok=True)

pause = np.zeros(
    (round(sr * PAUSE_SECONDS), channels),
    dtype=np.float32
)

with sf.SoundFile(
    str(final_local),
    mode="w",
    samplerate=sr,
    channels=channels,
    subtype="PCM_16",
) as output_file:

    for index, part in enumerate(parts):
        for audio_block in sf.blocks(
            str(part),
            blocksize=65536,
            dtype="float32",
            always_2d=True,
        ):
            output_file.write(audio_block)

        if index < len(parts) - 1:
            output_file.write(pause)

assert valid_wav(final_local)

# ============================================================
# 8. SAVE FINAL WAV TO DRIVE AND VERIFY
# ============================================================

DRIVE_ROOT.mkdir(parents=True, exist_ok=True)
final_drive = DRIVE_ROOT / "cua_so_thien_tap_01_full.wav"

if final_drive.exists():
    if checksum(final_drive) != checksum(final_local):
        raise FileExistsError(
            f"Different final WAV already exists: {final_drive}"
        )
    print("Final WAV already exists and matches.")
else:
    with final_local.open("rb") as src, final_drive.open("xb") as dst:
        shutil.copyfileobj(src, dst)

assert checksum(final_local) == checksum(final_drive)
assert valid_wav(final_drive)

print("\n" + "=" * 50)
print("VOICEOVER COMPLETED")
print("DRIVE SAVE VERIFIED")
print("Final WAV:", final_drive)
print("Chunks:", len(parts))
print("Duration:", round(sf.info(str(final_drive)).duration, 2), "seconds")
print("Job ID:", job_hash)
print("=" * 50)
```
